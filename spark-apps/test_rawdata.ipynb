```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, coalesce, lit

spark = (
    SparkSession.builder
    .appName("check-raw")
    .master("spark://spark-master:7077")
    .getOrCreate()
)

path = "hdfs://namenode:9000/data/raw/chicago_crimes.parquet"

df = spark.read.parquet(path)

# rows where at least one column is NOT null
not_all_null = df.filter(coalesce(*[col(c) for c in df.columns]).isNotNull())

# (optional) rows where ALL columns are null
all_null = df.filter(coalesce(*[col(c) for c in df.columns]).isNull())

print("Total rows:", df.count())
print("Not-all-null rows:", not_all_null.count())
print("All-null rows:", all_null.count())

# Show some valid rows
not_all_null.show(50, truncate=False)

print("Row count:", df.count())
```
